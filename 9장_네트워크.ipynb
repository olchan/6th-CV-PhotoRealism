## **최대 흐름 VS 최소 비용 흐름 문제**

| 항목       | 최대 흐름(Max Flow)              | 최소 비용 흐름(Min-Cost Flow)             |
| -------- | ---------------------------- | ----------------------------------- |
| 목적       | **흐름의 총량을 최대화**              | **주어진 수요량을 채우면서 비용을 최소화**           |
| 흐름 양     | 가능한 많이 흘리는 게 목적              | **지정된 양만큼만** 흘려야 함                  |
| 비용       | 간선에 비용 없음                    | **간선마다 비용(cost)** 있음                |
| 노드 수요/공급 | 소스(source) → 싱크(sink) 1:1 관계 | **모든 노드에 공급/수요(balanced flow)가 존재** |
| 제약       | 용량만 고려                       | 용량 + 비용 + 수요조건까지 고려                 |

>> **"최소 비용 흐름 문제에서는 각 노드마다 흘리고 받을 흐름의 양이 정해져 있고,
각 간선마다 흘림의 비용과 용량 상한이 있다."**

## 용량 제한이 있는 네트워크 심플렉스 해법

"잔여 용량 네트워크에서 어떤 호(간선)가 용량까지 흐름이 찼을 경우, 왜 그 호를 반대 방향으로 바꾸고 비용 부호를 음수로 바꾸는가?"

>> 현재 흐름 f(u→v)가 c(u→v)에 도달하면 더 이상 흐름을 정방향으로 추가할 수 없음. 하지만 네트워크 심플렉스는 다음 반복에서 흐름을 감소시키는 것도 허용해야 함 (즉, 되돌릴 수 있어야 함) ➡️ 그래서 역방향 간선 v→u를 잔여 네트워크에 추가하는 것

>> **역방향 간선의 비용은 왜 부호를 바꾸는가?**
네트워크 심플렉스는 최소 비용 흐름(min-cost flow) 문제를 다루고 있는데, u→v 방향의 흐름을 줄인다는 건, 사실상 v→u 방향으로 흐름을 보낸 것과 같다. 흐름을 되돌리면 그만큼 **이득(비용 감소)**이 있는 것처럼 해석됨
