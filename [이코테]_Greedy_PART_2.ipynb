# Greedy
: 매 순간 가장 좋아보이는 것을 선택하며, 현재의 선택이 나중에 미칠 영향에 대해서는 고려하지 않는다.

- 기준에 따라 좋은 것을 선택하는 알고리즘이기에,**가장 큰 순서대로, 가장 작은 순서대로와 같은 기준을 알게 모르게 제시**해준다.

```python
# 거스름돈을 주는 최소 동전 개수
n = 1260

count = 0

coin = [500, 100, 50, 10]

for i in coin:
    if ( n // i > 0 ):
        count += n // i
        n = n % i

print(count)

# 이 문제가 최적해를 보장하는 이유는 겂아 큰 동전이 항상 그보다 작은 동전의 배수 형태이기 때문
# Greedy 문제에서는 문제 풀이를 위한 아이디어를 떠올렸다면,
# 이 아이디어가 정당한지를 검토할 수 있어야 한다.
```
